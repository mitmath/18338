```julia
using LinearAlgebra
```

```julia
using Plots
```

```julia
sym(A)=Hermitian(A+A')/sqrt(8)
n=80
s=1000
h=(0:10*s)/s
```

```julia
G= [ sym(randn(n,n)+im*randn(n,n)) for i = h ] / √(s)
G[1]*=0;
G=cumsum(G)
e=map(eigmax,G)
plot()
for j=1:n  
    plot!(h, [ e[i]/sqrt(i/s) for i=1:10*s+1 ])  
end
```
